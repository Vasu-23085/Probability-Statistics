```python
# Program for probability analysis of Binomial,
# Poisson, and Normal distributions

from scipy.stats import binom, poisson, norm

# -------------------------------
# Binomial Distribution
# -------------------------------
n = 10 # number of trials
p = 0.5 # probability of success
x = 4 # number of successes
binomial_prob = binom.pmf(x, n, p)
print("BINOMIAL DISTRIBUTION")
print("----------------------")
print("Probability P(X=4) =", binomial_prob)
print("Mean =", n * p)
print("Variance =", n * p * (1 - p))

# -------------------------------
# Poisson Distribution
# -------------------------------
lam = 3 # average rate
x = 2
poisson_prob = poisson.pmf(x, lam)
print("\nPOISSON DISTRIBUTION")
print("----------------------")
print("Probability P(X=2) =", poisson_prob)
print("Mean =", lam)
print("Variance =", lam)

# -------------------------------
# Normal Distribution
# -------------------------------
mu = 50
sigma = 10
x = 60
normal_prob = norm.pdf(x, mu, sigma)
print("\nNORMAL DISTRIBUTION")
print("----------------------")
print("Probability Density at x=60 =", normal_prob)
print("Mean =", mu)
print("Variance =", sigma ** 2)
```
