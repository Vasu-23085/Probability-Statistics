```python
import numpy as np
from scipy.stats import skew, kurtosis

# Input database
data = np.array([100,123,457,690,723,624,1040])
print("Dataset : " , data)

# Mean
mean = np.mean(data)
print("\nMean = ", mean)

# Central Movements
mu2 = np.mean((data - mean) ** 2);
mu3 = np.mean((data - mean) ** 3);
mu4 = np.mean((data - mean) ** 4);
print("\nSecond Central Movement (Variance) =", mu2);
print("\nThird Central Movement =", mu3)
print("\nFourth Central Movement =", mu4)

# Skewness
skewness = skew(data)
print("\nSkewness =", skewness)

# Kurtosis
kurt = kurtosis(data, fisher=False)
print("\nKurtosis =", kurt)

# Interpretation
if skewness > 0:
  print("\nDistribution is positively skewed")
elif skewness < 0:
  print("\nDistribution is negetively skewed")
else:
  print("\nDistribution is symmetric")
if kurt > 3:
  print("\nDistribution is leptokurtic")
elif kurt < 3:
  print("\nDistribution is platykurtic")
else:
  print("\n Distribution is Mesokurtic")



```
