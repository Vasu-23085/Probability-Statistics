```python
# Program to calculate probabilities using Bayes' Theorem

# Given probabilities
P_D = 0.01 # Probability of disease
P_Pos_given_D = 0.99

# Probability of positive test if diseased
P_Pos_given_notD = 0.05

# Probability of positive test if not diseased
# Probability of not having disease
P_notD = 1 - P_D

# Total probability of positive test
P_Pos = (P_Pos_given_D * P_D) + (P_Pos_given_notD * P_notD)

# Bayes theorem
P_D_given_Pos = (P_Pos_given_D * P_D) / P_Pos

# Display results
print("Probability of Disease =", P_D)
print("Probability of Positive Test =", P_Pos)
print("Probability of Disease given Positive Test =", P_D_given_Pos)
```
